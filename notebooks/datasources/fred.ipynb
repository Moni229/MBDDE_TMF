```python
import base64
# ==========================================
# Imports
# ==========================================

from datetime import datetime

from src.sources.fred import FredSource
from src.utils.paths import get_landing_paths
from src.writer.landing_writer import LandingWriter

# ==========================================
# Configuración
# ==========================================
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()
secret = w.secrets.get_secret(scope='macroeconomy', key='FRED-API-KEY-2')
api_key = base64.b64decode(secret.value).decode("utf-8")

print(api_key)
SERIES = "CPIAUCSL"

# ==========================================
# Lectura
# ==========================================

fred = FredSource(api_key)

data = fred.read(SERIES)

# ==========================================
# Escritura en Landing
# ==========================================
landing_paths = get_landing_paths()
writer = LandingWriter(landing_paths["fred"])
writer.write(
    data=data,
    source="fred",
    dataset="CPIAUCSL",
    timestamp=datetime.now()
)

print("Ingesta finalizada correctamente.")
```
