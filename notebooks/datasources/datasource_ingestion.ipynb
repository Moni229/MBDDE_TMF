```python
from macroeconomy.pipelines.landing_pipeline import LandingPipeline
from macroeconomy.utils.config import load_sources
from macroeconomy.writers.landing_writer import LandingWriter
from databricks.sdk import WorkspaceClient
import base64

w = WorkspaceClient()
secret = w.secrets.get_secret(scope='macroeconomy', key='FRED-API-KEY-2')
api_key = base64.b64decode(secret.value).decode("utf-8")


config = load_sources()

writer = LandingWriter()

pipeline = LandingPipeline(writer)
```

```python
from macroeconomy.sources.fred_source import FredSource
pipeline.run(
    FredSource(api_key),
    config["fred"]
)
```

```python
from macroeconomy.sources.eurostat_source import EurostatSource
pipeline.run(
    EurostatSource(),
    config["eurostat"]
)
```

```python
from macroeconomy.sources.yahoo_source import YahooFinanceSource
pipeline.run(
    YahooFinanceSource(),
    config["yahoo"]
)
```
