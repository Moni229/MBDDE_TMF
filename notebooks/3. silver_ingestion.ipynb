```python
tables = spark.sql("SHOW TABLES IN macroeconomy_bronze").collect()

for table in tables:
    table_name = table.tableName

    if table_name.startswith("fred_"):
        spark.sql(f"DROP TABLE IF EXISTS macroeconomy_silver.`{table_name}`")
        print(f"Eliminada: {table_name}")
```
